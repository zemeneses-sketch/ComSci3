## Interacting objects

### Lesson 2-2 - Hero RPG

```python
class Hero:
  def __init__(self, name, hp):
    self.name = name
    self.hp = hp

  def take_damage(self, damage):
    self.hp -= damage

Arthur = Hero("Arthur", 100)
Morgana = Hero("Morgana", 100)

Arthur.take_damage(10)
print(f"{Arthur.name}'s Final HP is: {Arthur.hp}")
print(f"{Morgana.name}'s Final HP is: {Morgana.hp}")

```
